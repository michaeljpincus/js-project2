```python
import glob 
from PIL import Image 

image_files = sorted(glob.glob('shadows/*.png'))

frames = [Image.open(image) for image in image_files]

frames[0].save(
    'animated.shadows.gif',
    format='GIF',
    save_all=True,
    append_images=frames[1:],
    duration=500,
    loop=0
)
```
